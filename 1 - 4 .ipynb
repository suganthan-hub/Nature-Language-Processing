```python
import re

text = "My email is student123@gmail.com and my phone number is 9876543210."

# Search for an email pattern
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'

# Search for phone number
phone_pattern = r'\b\d{10}\b'

email = re.search(email_pattern, text)
phone = re.search(phone_pattern, text)

if email:
    print("Email found:", email.group())
else:
    print("Email not found")

if phone:
    print("Phone number found:", phone.group())
else:
    print("Phone number not found")
```

```python
def finite_automaton(string):
    state = 'q0'

    for char in string:
        if state == 'q0':
            if char == 'a':
                state = 'q1'
            else:
                state = 'q0'

        elif state == 'q1':
            if char == 'b':
                state = 'q2'
            elif char == 'a':
                state = 'q1'
            else:
                state = 'q0'

        elif state == 'q2':
            if char == 'a':
                state = 'q1'
            else:
                state = 'q0'

    return state == 'q2'


# Test strings
strings = ["ab", "aab", "cab", "abc", "abab", "helloab"]

for s in strings:
    if finite_automaton(s):
        print(s, "-> Accepted")
    else:
        print(s, "-> Rejected")
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk import pos_tag

text = "The students are studying artificial intelligence."

# Tokenization
words = word_tokenize(text)

# Stemming
stemmer = PorterStemmer()

print("Word\t\tStem\t\tPOS")
print("--------------------------------")

for word in words:
    if word.isalpha():
        stem = stemmer.stem(word)
        pos = pos_tag([word])[0][1]
        print(f"{word:<15}{stem:<15}{pos}")
```

```python
def plural(noun):
    # State q0: start with noun
    state = 'q0'

    # Rule 1: words ending in s, x, z, ch, sh → add "es"
    if noun.endswith(('s', 'x', 'z', 'ch', 'sh')):
        state = 'q1'
        return noun + 'es'

    # Rule 2: consonant + y → replace y with ies
    elif noun.endswith('y') and len(noun) > 1 and noun[-2] not in 'aeiou':
        state = 'q2'
        return noun[:-1] + 'ies'

    # Rule 3: normal nouns → add s
    else:
        state = 'q3'
        return noun + 's'


# Test nouns
nouns = ["cat", "dog", "bus", "box", "church", "baby", "book"]

for noun in nouns:
    print(noun, "->", plural(noun))
```
