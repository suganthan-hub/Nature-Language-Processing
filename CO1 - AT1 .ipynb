```python
import re

# Sample resume data
resume = """
Name: Suganthan Kumaresan
Email: suganthan@gmail.com
Mobile: +91 9876543210
Skills: Python, Java, SQL, Machine Learning, NLP
Experience: 3 years
"""

# 1. Extract candidate name
name = re.search(r"Name\s*:\s*([A-Za-z ]+)", resume)

# 2. Extract email
email = re.search(r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}", resume)

# 3. Extract mobile number
mobile = re.search(r"(?:\+91[\s-]?)?[6-9]\d{9}", resume)

# 4. Detect technical skills
skills_list = ["Python", "Java", "SQL", "Machine Learning", "NLP"]
found_skills = []

for skill in skills_list:
    if re.search(r"\b" + re.escape(skill) + r"\b", resume, re.IGNORECASE):
        found_skills.append(skill)

# 5. Extract years of experience
experience = re.search(r"(\d+)\s*(?:years?|yrs?)\s*(?:of)?\s*experience", resume, re.IGNORECASE)

# Display extracted information
candidate_name = name.group(1).strip() if name else "Not Found"
candidate_email = email.group() if email else "Not Found"
candidate_mobile = mobile.group() if mobile else "Not Found"
candidate_experience = int(experience.group(1)) if experience else 0

print("========== CANDIDATE PROFILE ==========")
print("Name        :", candidate_name)
print("Email       :", candidate_email)
print("Mobile      :", candidate_mobile)
print("Skills      :", ", ".join(found_skills))
print("Experience  :", candidate_experience, "years")

# Eligibility criteria
if candidate_experience >= 2 and "Python" in found_skills:
    print("\nEligibility : ELIGIBLE")
else:
    print("\nEligibility : NOT ELIGIBLE")
```

```python
import re

# Product list
products = [
    "Apple iPhone 15",
    "Samsung Galaxy S24",
    "Dell Laptop",
    "HP Laptop",
    "Lenovo Laptop",
    "Apple MacBook Air",
    "Samsung Galaxy Tab",
    "Sony Headphones",
    "Boat Bluetooth Speaker",
    "Logitech Wireless Mouse"
]

print("========== PRODUCT SEARCH SYSTEM ==========")

# Get search keyword from user
keyword = input("Enter search keyword: ")

# 1. Exact keyword search
exact_matches = [
    product for product in products
    if re.search(r"\b" + re.escape(keyword) + r"\b", product)
]

print("\n1. EXACT KEYWORD SEARCH")
print("Matches:", exact_matches)

# 2. Prefix search
prefix_matches = [
    product for product in products
    if re.search(r"\b" + re.escape(keyword), product, re.IGNORECASE)
]

print("\n2. PREFIX SEARCH")
print("Matches:", prefix_matches)

# 3. Suffix search
suffix_matches = [
    product for product in products
    if re.search(re.escape(keyword) + r"\b", product, re.IGNORECASE)
]

print("\n3. SUFFIX SEARCH")
print("Matches:", suffix_matches)

# 4. Partial keyword search
partial_matches = [
    product for product in products
    if re.search(re.escape(keyword), product, re.IGNORECASE)
]

print("\n4. PARTIAL KEYWORD SEARCH")
print("Matches:", partial_matches)

# 5. Case-insensitive search
case_matches = [
    product for product in products
    if re.search(re.escape(keyword), product, re.IGNORECASE)
]

print("\n5. CASE-INSENSITIVE SEARCH")
print("Matches:", case_matches)

# Final report
print("\n========== SEARCH REPORT ==========")
print("Exact matches          :", len(exact_matches))
print("Prefix matches         :", len(prefix_matches))
print("Suffix matches         :", len(suffix_matches))
print("Partial matches        :", len(partial_matches))
print("Case-insensitive matches:", len(case_matches))
```

```python
import re

print("========== UNIVERSITY REGISTRATION SYSTEM ==========")

# Get student details
register_no = input("Enter Register Number: ")
email = input("Enter Institutional Email: ")
course_code = input("Enter Course Code: ")
semester = input("Enter Semester (1-8): ")
mobile = input("Enter Mobile Number: ")

# Validation patterns

# Register number: Example 23AID001
register_pattern = r"^\d{2}[A-Z]{2,4}\d{3}$"

# Institutional email
email_pattern = r"^[A-Za-z0-9._%+-]+@saveetha\.ac\.in$"

# Course code: Example CSE101, AIML201
course_pattern = r"^[A-Z]{2,5}\d{3}$"

# Semester: 1 to 8
semester_pattern = r"^[1-8]$"

# Indian mobile number
mobile_pattern = r"^[6-9]\d{9}$"

# Validate Register Number
if re.fullmatch(register_pattern, register_no):
    print("Register Number : Valid")
    register_valid = True
else:
    print("Register Number : Invalid")
    register_valid = False

# Validate Email
if re.fullmatch(email_pattern, email):
    print("Institutional Email : Valid")
    email_valid = True
else:
    print("Institutional Email : Invalid")
    email_valid = False

# Validate Course Code
if re.fullmatch(course_pattern, course_code):
    print("Course Code : Valid")
    course_valid = True
else:
    print("Course Code : Invalid")
    course_valid = False

# Validate Semester
if re.fullmatch(semester_pattern, semester):
    print("Semester : Valid")
    semester_valid = True
else:
    print("Semester : Invalid")
    semester_valid = False

# Validate Mobile
if re.fullmatch(mobile_pattern, mobile):
    print("Mobile Number : Valid")
    mobile_valid = True
else:
    print("Mobile Number : Invalid")
    mobile_valid = False

# Final registration status
print("\n========== FINAL REGISTRATION REPORT ==========")

if all([
    register_valid,
    email_valid,
    course_valid,
    semester_valid,
    mobile_valid
]):
    print("Registration Status : SUCCESSFUL")
    print("All student details are valid.")
else:
    print("Registration Status : FAILED")
    print("Please correct the invalid fields.")
```
