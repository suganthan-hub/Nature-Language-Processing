```python
import re
from collections import Counter

text = """Python is easy to learn. Python is a popular programming language.
Students use Python for data science and AI."""

# Split into sentences
sentences = re.split(r'[.!?]+', text)

print("Sentences:")
for sentence in sentences:
    if sentence.strip():
        print(sentence.strip())

# Split into words
words = re.findall(r'\b\w+\b', text.lower())

print("\nWords:")
print(words)

# Count words
word_count = Counter(words)

print("\nWord Count:")
print(word_count)

# Find most frequent word
most_frequent = word_count.most_common(1)[0]

print("\nMost Frequent Word:")
print(most_frequent[0], "->", most_frequent[1], "times")
```

```python
import re

text = """
My name is Suganthan Kumar.
My email is student@gmail.com.
You can contact me at 9876543210.
The meeting is on 07/10/2026.
Python and Artificial Intelligence are useful.
"""

# Find date
date_pattern = r'\b\d{2}/\d{2}/\d{4}\b'
dates = re.findall(date_pattern, text)

# Find mobile numbers
mobile_pattern = r'\b[6-9]\d{9}\b'
mobiles = re.findall(mobile_pattern, text)

# Find email IDs
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
emails = re.findall(email_pattern, text)

# Find capitalized words
capital_pattern = r'\b[A-Z][a-zA-Z]*\b'
capital_words = re.findall(capital_pattern, text)

print("Dates:")
print(dates)

print("\nMobile Numbers:")
print(mobiles)

print("\nEmail IDs:")
print(emails)

print("\nCapitalized Words:")
print(capital_words)
```

```python
def finite_automaton(string):
    state = 'q0'
    path = [state]

    for ch in string:

        if state == 'q0':
            if ch == 'a':
                state = 'q1'
            else:
                state = 'q0'

        elif state == 'q1':
            if ch == 'a':
                state = 'q1'
            else:
                state = 'q2'

        elif state == 'q2':
            if ch == 'a':
                state = 'q1'
            else:
                state = 'q0'

        path.append(state)

    if state == 'q2':
        result = "ACCEPTED"
    else:
        result = "REJECTED"

    return path, result


# Test strings
strings = ["ab", "aab", "babab", "aba", "abb", "ba"]

for string in strings:
    path, result = finite_automaton(string)

    print("Input:", string)
    print("Path :", " -> ".join(path))
    print("Result:", result)
    print()
```

```python
strings = ["python", "programming", "python programming", "AI"]

for s in strings:
    print("String:", s)

    print("Is empty? :", "Yes" if s == "" else "No")
    print("Starts with 'p'? :", "Yes" if s.lower().startswith("p") else "No")
    print("Ends with 'g'? :", "Yes" if s.lower().endswith("g") else "No")
    print("Contains 'python'? :", "Yes" if "python" in s.lower() else "No")

    print()
```
